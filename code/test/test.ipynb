```python
import os
prefix = "graphSage_"
directory = "../models/ego"
for filename in os.listdir(directory):
    new_filename = prefix + filename
    old_path = os.path.join(directory, filename)
    new_path = os.path.join(directory, new_filename)
    os.rename(old_path, new_path)
    print(f"Renamed {filename} to {new_filename}")
```

```python
import pandas as pd
import numpy as np

# 定义df1
df1 = pd.DataFrame({'alpha':['A','B','C','D','E'],
					'feature1':[1,1,2,3,3],
					'feature2':['low','medium','medium','high','low']})

df2 = pd.DataFrame({'alpha':['A','B','Z'],
					'feature1':[1,1,4],
					'feature2':['low','medium','high']
					})

df2 = pd.DataFrame({'alpha':['A','B','Z'],
					'feature1':[1,1,4]
					})


df3 = pd.concat([df1,df2])
# 去除重复项
df3 = df3.drop_duplicates()
print(df1,df1.index)
print(df2,df1.index)
print(df3)
```

```python
# 单列的左连接
# 定义df1
df1 = pd.DataFrame({'alpha':['A','B','B','C','D','E'],
					'feature1':[1,1,2,3,3,1],
    				'feature2':['low','medium','medium','high','low','high']})
# 定义df2
df2 = pd.DataFrame({'alpha':['A','A','B','F'],
					'pazham':['apple','orange','pine','pear'],
                    'kilo':['high','low','high','medium'],
                    'price':np.array([5,6,5,7])})
# 基于共同列alpha的左连接
df5 = pd.merge(df1,df2,how='left',on='alpha')

print(df1)
print(df2)
print(df5)
```
